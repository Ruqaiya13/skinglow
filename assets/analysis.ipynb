```python
import requests
import os
import json

# 1. تحميل البيانات من ملف JSON
with open('skin_data.json', 'r', encoding='utf-8') as file:
    data = json.load(file)

# 2. تنزيل الصور وحفظها في مجلدات حسب نوع البشرة
for item in data:
    try:
        image_url = item["image_url"]
        skin_type = item["skin_type"]
        
        # تجاهل العناصر التي ليس لديها نوع بشرة محدد
        if not skin_type:
            print(f"تحذير: نوع البشرة غير محدد للصورة {image_url}")
            continue
        
        # إنشاء مجلد لنوع البشرة إذا لم يكن موجودًا
        os.makedirs(f"skin_dataset/{skin_type}", exist_ok=True)
        
        # تنزيل الصورة
        response = requests.get(image_url)
        response.raise_for_status()  # التحقق من نجاح الطلب
        
        # استخراج اسم الملف من الرابط
        filename = image_url.split("/")[-1].split("?")[0]  # إزالة البارامترات
        
        # حفظ الصورة
        with open(f"skin_dataset/{skin_type}/{filename}", "wb") as img_file:
            img_file.write(response.content)
            
        print(f"تم تنزيل {filename} وحفظه في مجلد {skin_type}")
        
    except Exception as e:
        print(f"فشل في تنزيل {image_url}: {str(e)}")
```

```python
import requests
import os
import json

# تحميل البيانات من ملف JSON
with open('skin_data2.json', 'r', encoding='utf-8') as file:
    data = json.load(file)

# إنشاء ملف لتسجيل الأخطاء
failed_downloads = []
with open('failed_downloads.txt', 'w', encoding='utf-8') as log_file:
    log_file.write("قائمة بالروابط الفاشلة:\n")

for item in data:
    try:
        image_url = item["image_url"]
        skin_type = item["skin_type"]
        
        if not skin_type:
            print(f"تحذير: نوع البشرة غير محدد للصورة {image_url}")
            continue
        
        os.makedirs(f"skin_dataset/{skin_type}", exist_ok=True)
        filename = image_url.split("/")[-1].split("?")[0]
        
        # إضافة User-Agent لتجنب الحظر
        headers = {
            'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.3'
        }
        
        response = requests.get(image_url, headers=headers, timeout=10)
        response.raise_for_status()
        
        with open(f"skin_dataset/{skin_type}/{filename}", "wb") as img_file:
            img_file.write(response.content)
            
        print(f"تم تنزيل: {filename}")
        
    except Exception as e:
        print(f"فشل في تنزيل {image_url}: {str(e)}")
        failed_downloads.append(image_url)
        with open('failed_downloads.txt', 'a', encoding='utf-8') as log_file:
            log_file.write(f"{image_url} | السبب: {str(e)}\n")

print(f"تم تنزيل {len(data) - len(failed_downloads)} من أصل {len(data)} صورة.")
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, Flatten, Dense

# إنشاء مولدات البيانات مع class_mode='categorical'
train_datagen = ImageDataGenerator(rescale=1./255)
val_datagen = ImageDataGenerator(rescale=1./255)

train_generator = train_datagen.flow_from_directory(
    r'C:\Users\user\skinglow\assets\skin_dataset\train',
    target_size=(150, 150),
    batch_size=32,
    class_mode='categorical'
)

val_generator = val_datagen.flow_from_directory(
    r'C:\Users\user\skinglow\assets\skin_dataset\validation',
    target_size=(150, 150),
    batch_size=32,
    class_mode='categorical'
)

# بناء النموذج
model = Sequential([
    Input(shape=(150, 150, 3)),
    Conv2D(32, (3,3), activation='relu'),
    MaxPooling2D(2,2),
    Conv2D(64, (3,3), activation='relu'),
    MaxPooling2D(2,2),
    Conv2D(128, (3,3), activation='relu'),
    MaxPooling2D(2,2),
    Flatten(),
    Dense(512, activation='relu'),
    Dense(3, activation='softmax')
])

# تجميع النموذج
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# تدريب النموذج
history = model.fit(
    train_generator,
    epochs=30,
    validation_data=val_generator
)
```

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator

# 1. تعريف مسار دليل الاختبار
test_dir = r'C:\Users\user\skinglow\assets\skin_dataset\test'  # غيّر هذا إلى المسار الصحيح

# 2. تهيئة ImageDataGenerator للاختبار
test_datagen = ImageDataGenerator(rescale=1./255)

# 3. إنشاء test_generator
test_generator = test_datagen.flow_from_directory(
    test_dir,
    target_size=(150, 150),  # يجب أن تطابق الأبعاد المستخدمة في التدريب
    batch_size=32,
    class_mode='categorical',  # أو 'binary' حسب مشكلتك
    shuffle=False  # مهم للحفاظ على ترتيب البيانات عند التقييم
)

# 4. تقييم النموذج
test_loss, test_acc = model.evaluate(test_generator)
print(f'الدقة النهائية: {test_acc*100:.2f}%')
```
